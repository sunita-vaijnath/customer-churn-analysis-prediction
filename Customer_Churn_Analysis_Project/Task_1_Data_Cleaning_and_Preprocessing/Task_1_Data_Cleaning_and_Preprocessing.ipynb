their subscriptions (churn). Before any churn driver can be analyzed or predicted, the raw
operational data must be audited and cleaned, because analysis built on unverified data produces

Audit the raw Telco Customer Churn dataset for missing values, duplicates, incorrect datatypes and

for downstream modeling, and output a cleaned dataset that Tasks 2–4 can rely on.

## 6. Data Quality Investigation

### 6.1 Missing values (standard nulls)

```python
missing_counts = df.isnull().sum()
missing_pct = (missing_counts / len(df) * 100).round(2)
missing_summary = pd.DataFrame({'missing_count': missing_counts, 'missing_pct': missing_pct})
missing_summary[missing_summary['missing_count'] > 0].sort_values('missing_count', ascending=False)
```

**Observation:** A standard `.isnull()` check reports **0 missing values** across all 21
columns. However, this does not necessarily mean the data is complete — some columns may contain
"hidden" missing values represented as blank strings or placeholder text rather than true `NaN`.
`TotalCharges` is investigated specifically below because it is loaded as a text column, which is
unusual for a monetary field.

### 6.2 Duplicate rows and duplicate customer IDs

```python
duplicate_rows = df.duplicated().sum()
duplicate_ids = df['customerID'].duplicated().sum()

print(f"Fully duplicated rows: {duplicate_rows}")
print(f"Duplicate customerID values: {duplicate_ids}")
print(f"Unique customerID count: {df['customerID'].nunique()} out of {len(df)} rows")
```

**Observation:** There are **0 fully duplicated rows** and **0 duplicate `customerID`**
values — every row represents a unique customer. No duplicate-removal treatment is required.

### 6.3 `TotalCharges` datatype investigation

```python
print("Current dtype of TotalCharges:", df['TotalCharges'].dtype)

# Attempt numeric conversion and see which values fail to convert
totalcharges_numeric_attempt = pd.to_numeric(df['TotalCharges'], errors='coerce')
non_numeric_mask = totalcharges_numeric_attempt.isnull()

print(f"\nRows where TotalCharges cannot be converted to a number: {non_numeric_mask.sum()}")
df.loc[non_numeric_mask, ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']]
```

**Observation:** `TotalCharges` is stored as a text (`object`) column instead of a number.
When forced to numeric, **11 rows** fail to convert. Inspecting those rows shows their raw
`TotalCharges` value is an **empty string** (whitespace), and in every one of these 11 rows
**`tenure` = 0**, meaning the customer had just signed up and had not yet been billed a first full
cycle — so no total-charges figure existed yet at the time of data extraction. This is a logical,
explainable data-entry gap rather than a random data-quality error, and all 11 of these customers
have `Churn = No`.

**Why conversion cannot be done blindly:** simply calling `pd.to_numeric()` without first checking
would silently convert these blank strings to `NaN`, which is correct — but the investigation above
is required to confirm the cause is genuine "new customer, not yet billed" rather than a data-entry
error which might warrant a different treatment (e.g. supplying a default value or excluding the
customer entirely).

## 7. Missing-Value Treatment

```python
# Convert TotalCharges to numeric now that the cause of the blank values is understood
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

print(f"Missing values in TotalCharges after conversion: {df['TotalCharges'].isnull().sum()}")
print(f"These correspond to customers with tenure = 0: "
      f"{(df.loc[df['TotalCharges'].isnull(), 'tenure'] == 0).all()}")
```

**Decision:** For these 11 zero-tenure customers, `TotalCharges` is logically **0** — they
have not been billed yet, so total historical charges are genuinely zero rather than "unknown".
Imputing with the column mean/median would fabricate a charge history these customers never had, and
dropping them would discard 11 real, valid customer records for a segment (new sign-ups) that is
analytically important for churn analysis. Therefore, the correct treatment is to **impute these 11
values with 0**, not with a mean/median and not by row deletion.

```python
df['TotalCharges'] = df['TotalCharges'].fillna(0)

print(f"Missing values in TotalCharges after treatment: {df['TotalCharges'].isnull().sum()}")
print(f"TotalCharges dtype: {df['TotalCharges'].dtype}")
df['TotalCharges'].describe()
```

## 11. Should `customerID` Be Used as a Model Feature?

`customerID` is a unique alphanumeric identifier (verified above: 7,043 unique values for
7,043 rows, zero duplicates). It carries **no predictive signal** about customer behavior — it is
an arbitrary label generated at account creation. Including it as a model feature would:

- Add noise / a spurious high-cardinality feature that provides no real information,
- Risk the model "memorizing" IDs rather than learning genuine churn patterns,
- Provide zero business interpretability.

**Decision:** `customerID` is retained in the dataset **only as a lookup/reference key** (used in
Task 5 to identify specific high-risk customers) and is **excluded from all model features** in
Task 4.

## 12. Categorical Encoding

Encoding strategy:

- **Binary Yes/No columns** (`Partner`, `Dependents`, `PhoneService`, `PaperlessBilling`, `Churn`)
  → mapped to `1`/`0`, since two categories with a natural true/false meaning do not need extra
  dummy columns.
- **`gender`** (`Male`/`Female`) → mapped to a binary `0`/`1` flag for the same reason.
- **`SeniorCitizen`** → already numeric `0`/`1`, left as-is.
- **Multi-category columns** (`MultipleLines`, `InternetService`, `OnlineSecurity`, `OnlineBackup`,
  `DeviceProtection`, `TechSupport`, `StreamingTV`, `StreamingMovies`, `Contract`, `PaymentMethod`)
  → **one-hot encoded**, because these have 3+ unrelated categories (including "No internet
  service"/"No phone service" as a genuinely distinct category, not a missing value) and have no
  inherent order, so a single integer mapping would falsely imply rank/order between categories.
- **`customerID`** → excluded from encoding entirely (see Section 11).

A separate copy of the dataframe is encoded so that the human-readable cleaned dataset (saved at the
end of this notebook) still shows meaningful category names for reporting purposes; the fully
one-hot-encoded numeric version is produced as part of the Task 4 modeling pipeline instead, to
avoid duplicating large sparse column sets across every task's file.

```python
binary_yes_no_cols = ['Partner', 'Dependents', 'PhoneService', 'PaperlessBilling', 'Churn']
for col in binary_yes_no_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})

print("Binary encoding applied to:", binary_yes_no_cols + ['gender'])
df[binary_yes_no_cols + ['gender', 'SeniorCitizen']].head()
```

The remaining multi-category columns are intentionally **left as text in this cleaned
dataset** (rather than one-hot encoded here) so that Tasks 2 and 3 (EDA and segmentation) can group
and plot by readable category names such as `"Month-to-month"` or `"Fiber optic"`. The one-hot
encoding required specifically for the machine-learning model is applied inside a `ColumnTransformer`
in Task 4, which is the standard, leakage-safe place to do it (fit only on the training split).

## 15. Task 1 Conclusion

The raw dataset was structurally very clean: there were **zero duplicate rows**, **zero
duplicate customer IDs**, and **zero missing values under a naive null check**. The single genuine
data-quality issue was `TotalCharges` being stored as text due to **11 blank-string values**, all
of which belonged to customers with `tenure = 0` (brand-new sign-ups not yet billed). These were
corrected by converting the column to numeric and imputing the 11 blanks with `0`, which reflects
their true billing history rather than an unknown value.

Binary categorical columns (`Partner`, `Dependents`, `PhoneService`, `PaperlessBilling`, `Churn`,
`gender`) were mapped to `1`/`0`. Multi-category service and billing columns were left as readable
text for EDA/segmentation and will be one-hot encoded inside the modeling pipeline in Task 4 to avoid
data leakage. `customerID` was confirmed to be a unique, non-predictive identifier and will be
excluded from all model features.

The cleaned dataset (`Task_1_Cleaned_Dataset.csv`, 7,043 rows × 21 columns, 0 missing values, 0
duplicates) is now ready to support the Exploratory Data Analysis (Task 2), Customer Segmentation
(Task 3), and Churn Prediction Model (Task 4) that follow.
